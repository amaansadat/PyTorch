```python
import torch
```

```python
#to find the derivative of a tensor, set requires_grad=true
x=torch.tensor(3.0,requires_grad=True)
y=x**2
```

```python
x
```

```python
y
```

```python
y.backward()
```

```python
x.grad
```

```python
import math
def dz_dx(x):
  return 2*x*math.cos(x**2)

```

```python
dz_dx(4)
```

```python
#Now calculation derivative of tensor at x=4 in a different way than above 2 cells i.e by using autograd function
x=torch.tensor(4.0,requires_grad=True)
y=x**2
z=torch.sin(y)
```

```python
x
```

```python
y
```

```python
z
```

```python
z.backward()
```

```python
x.grad
```
