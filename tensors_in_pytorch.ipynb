MATHEMATICAL OPERATIONS IN TENSORS


### 1. Scalar operation

```python
x = torch.rand(2,2)
x
```

```python
#on adding a scalar: it'll be added in each element
x+2
#lly subtraction,multiplication,division
#int division
(x*100)//2
# mod
((x * 100)//2)%2
# power
x**2
```

### 2. Element wise operation

```python
a = torch.rand(2,3)
b = torch.rand(2,3)

print(a)
print(b)
```

```python
#add
a + b
#subtract
a - b
#multiply
a * b
#division
a / b
#power
a ** b
#mod
a % b
```

```python
c = torch.tensor([1, -2, 3, -4])
```

```python
# abs
torch.abs(c)
```

```python
# negative
torch.neg(c)
```

```python
d = torch.tensor([1.9, 2.3, 3.7, 4.4])
```

### 3. Reduction operation

```python
e = torch.randint(size=(2,3), low=0, high=10, dtype=torch.float32)
e
```

```python
# sum of matrix
torch.sum(e)
# sum of columns
torch.sum(e, dim=0)
# sum of rows
torch.sum(e, dim=1)
```

```python
# mean(works only when datatype is float)
torch.mean(e)
# mean along col
torch.mean(e, dim=0)
```

```python
# argmax:tells position max element in tensor
torch.argmax(e)
```

```python
# argmin:tells position of min element in tensor
torch.argmin(e)
```

```python
# standard deviation
torch.std(e)
```

```python
# variance
torch.var(e)
```

### 4. Matrix operations

```python
f = torch.randint(size=(2,3), low=0, high=10)
g = torch.randint(size=(3,2), low=0, high=10)

print(f)
print(g)
```

```python
# matrix multiplcation
torch.matmul(f, g)
```

```python
vector1 = torch.tensor([1, 2])
vector2 = torch.tensor([3, 4])

# dot product
torch.dot(vector1, vector2)
```

```python
h = torch.randint(size=(3,3), low=0, high=10, dtype=torch.float32)
h
```

```python
# determinant
torch.det(h)
```

```python
# relu:returns tensor containing positive values only
torch.relu(h)
```

```python
# inverse
torch.inverse(h)
```

```python
import time
#taking size of matrices as 1000
size=1000
# Create random matrices on CPU
matrix_cpu1 = torch.randn(size, size)
matrix_cpu2 = torch.randn(size, size)

# Measure time on CPU
start_time = time.time()
result_cpu = torch.matmul(matrix_cpu1, matrix_cpu2)  # Matrix multiplication on CPU
cpu_time = time.time() - start_time

print(f"Time on CPU: {cpu_time:.4f} seconds")

# Move matrices to GPU
matrix_gpu1 = matrix_cpu1.to('cuda')
matrix_gpu2 = matrix_cpu2.to('cuda')

# Measure time on GPU
start_time = time.time()
result_gpu = torch.matmul(matrix_gpu1, matrix_gpu2)  # Matrix multiplication on GPU
torch.cuda.synchronize()  # Ensure all GPU operations are complete
gpu_time = time.time() - start_time

print(f"Time on GPU: {gpu_time:.4f} seconds")

# Compare results
print("\nSpeedup (CPU time / GPU time):", cpu_time / gpu_time)
```
