CREATING A TENSOR

```python
#Using empty function
torch.empty(2,3)
```

```python
#using zeros function to create a tensor of zeros
torch.zeros(2,3)
```

```python
#using ones function to create tenor of ones
torch.ones(2,3)
```

```python
#using rand function to create a tenor of random values bw 0 to 1
torch.rand(2,3)
```

```python
#each time when we run rand function then output will be different and to preserve the value of rand we define a seed.
#manual_seed
torch.manual_seed(100)
torch.rand(2,3)
#now whenever we use this seed(100) to use rand function then each time we'll get the same values
```

```python
torch.manual_seed(100)
torch.rand(2,3)
```

```python
#using tensor function to create a tensor of desired values
torch.tensor([[1,2,3],[4,5,6]])
```

```python
#using arange function, lly to numpy
torch.arange(0,10,2)
```

TENSOR SHAPES


```python
x=torch.tensor([[1,2,3],[4,5,6]])
x.shape
```

```python
#to create a tensor of same shape as given tensor use _like(giventensor) function i.e empty_like(x),ones_like(x) etc etc.
#And rand_like(x) will create same tensor only if given tensor has value bw 0 to 1 as rand function creates a tensor of value 0 to 1 so specify dtype=torch.float32
```

```python
torch.zeros_like(x)
```

```python
torch.rand_like(x, dtype=torch.float32)
```

TENSOR DATA TYPE

```python
#to find the datatype of tensor use x.dtype
x.dtype
```

```python
#to assign data type
torch.tensor([1.0,2.0,3.0], dtype=torch.int32)
```

```python
#to change the datatype of existing tensor use to() function
x.to(torch.float32)
```
